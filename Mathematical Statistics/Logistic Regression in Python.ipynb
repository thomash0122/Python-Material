# Logistic Regression in Python

## Introduction

In Python, we can use statsmodels for a detailed logistic regression model similar to R's glm(), or scikit-learn for a more machine-learning-oriented approach. Here, we'll use statsmodels to match R's output style.

## Prerequisities

Before we begin, ensure you have the following Python libraries installed:

## Importing Required Libraries

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from sklearn.model_selection import train_test_split
```

## Loading the Dataset

```python
df = pd.read_csv('PimaIndiansDiabetes2_dataset.csv')
```

### Drop NAs

```python
df.dropna(inplace=True)
```

## Data Splitting

```python
# Set seed for reproducibility
np.random.seed(123)

# Split data (80% train, 20% test)
train_data, test_data = train_test_split(df, test_size=0.2, stratify=df['diabetes'], random_state=123)
```

## Model Fitting

```python
# Prepare training data
X_train = sm.add_constant(train_data[['pregnant', 'glucose', 'pressure', 'triceps', 'insulin', 'mass', 'pedigree', 'age']])
y_train = train_data['diabetes'].map({'pos': 1, 'neg': 0})

# Fit the model
fit = sm.Logit(y_train, X_train).fit()

```

## Model Summary

```python
fit.summary()
```
