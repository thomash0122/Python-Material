## Probability Distribution in Python

This lesson is meant to teach how to find the probability of a given distribution in python.

We are going to go over the syntax of probability distribution in python.

Before do aything with the probability distribution, we must import the scipy.stats module to python:

```python
import scipy.stats
```

In general to perform any probability distirbution related function:

```python
import scipy.stats as stats

stats.<distribution_name>.<method()>
```

<distribution_name> can be:

* discrete:
    * bernoulli- A Bernoulli discrete random variable.
    * binom- A binomial discrete random variable.
    * geom- A geometric discrete random variable.
    * hypergeom- A hypergeometric discrete random variable.
    * logser- A Logarithmic (Log-Series, Series) discrete random variable.
    * poisson- A Poisson discrete random variable.
    * randint- A uniform discrete random variable.
    * and more...
    
* Continuous:
    * chi2- A chi-squared continuous random variable.
    * expon- An exponential continuous random variable.
    * f- An F continuous random variable.
    * gamma- A gamma continuous random variable.
    * logistic- A logistic (or Sech-squared) continuous random variable.
    * norm- A normal continuous random variable.
    * t- A Student's t continuous random variable.
    * uniform- A uniform continuous random variable.
    * and more...

<method()> can be: 


* rvs(df, loc=0, scale=1, size=1, random_state=None)- Generate random sample of a given distribution. returns an array
* pmf(parameters, loc=0)- Probability mass function. For discrete distribution
* pdf(parameters, loc=0)- Probability density function. For continuous distribution
* logpmf(parameters, loc=0)- Log of the probability mass function.
* cdf(parameters, loc=0)- Cumulative distribution function.
* logcdf(parameters, loc=0)- Log of the cumulative distribution function.
* sf(parameters, loc=0)- Survival function (also defined as 1 - cdf, but sf is sometimes more accurate).
* ppf(q, df, loc=0, scale=1)- Percent point function (inverse of cdf — percentiles). Used to find percentile
* stats(parameters, loc=0, moments=’mv’)- Mean(‘m’), variance(‘v’), skew(‘s’), and/or kurtosis(‘k’).* 
* expect(func, args=(p,), loc=0, lb=None, ub=None, conditional=False)- Expected value of a function (of one argument) with respect to the distribution.
* median(parameters, loc=0)- Median of the distribution.
* mean(parameters, loc=0)- Mean of the distribution.
* var(parameters, loc=0)- Variance of the distribution.
* std(parameters, loc=0)- Standard deviation of the distribution.
* interval(confidence, parameters, loc=0)- Confidence interval with equal areas around the median.

parameters refers to the parameters of the probability distribution

For example: If we want to plot a **normal** distribution with **mean = 8** and **standard deviation = 10**

```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.stats as stats

mean = 8 
std = 10

#define the domain of x
x = np.linspace(mean-3*std, mean+3*std,100)

#plot density curve
plt.plot(
    x,
    stats.norm.pdf(x, loc=8, scale=8)
)

#print graph
plt.show()
plt.close()
```
