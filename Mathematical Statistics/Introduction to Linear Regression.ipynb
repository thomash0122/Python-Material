### Construct Regression model

To construct a regression model in python, we need to create an instance of the OLS class

syntax:

```python
import statsmodels.api as sm
model = sm.OLS.from_formula("y ~ x", data = data_name)
```

### Create RegressionResult Object

```python
results = model.fit()
```

### Assumption of Linear Regression- Normality

The Normality assumption simply means the residual of the model must be normally distributed

The residual is given by
* actual value - fitted value

To check the normality assumption, we can simply plot the residual on the histogram plot.

If the residual is normally distributed, then the normality assumption holds

### Assumption of Linear Regression- Homoscedasticity

Homoscedasticity means **constant variance**

When a fitted value(x) vs. residual(y) scatter plot is plotted, if the plot:
* shows no sign of any pattern
* points are randomly spread across

then the homoscedasticity assumption holds
