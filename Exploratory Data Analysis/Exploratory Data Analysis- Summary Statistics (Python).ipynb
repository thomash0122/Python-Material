## Numerical Data Analysis

When summarizing numerical variables, we often want to describe:
* Central Location
* Spread

### Central Tendency for numerical variable

The central location (also called central tendency) is often used to communicate the “typical” value of a variable.

There are several common measures of central tendency:

* Mean
    * The average value of the variable, calculated as the sum of all values divided by the number of values.
* Median
    * The middle value of the variable when sorted.
* Mode
    * The most frequent value of the variable.
* Trimmed mean
    * The mean excluding x percent of the lowest and highest data points.

To find mean, median, mode, and Trimmed mean in python:

syntax:

```python
# Mean
    df_name.column_name.mean()

# Median
    df_name.column_name.median()

# Mode
    df_name.column_name.mode()

# Trimmed mean: we need to use the function: trim_mean() from the module: scipy.stats
    from scipy.stats import trim_mean
    trim_mean(df_name.column_name, proportiontocut=...%)
```

For our rentals DataFrame with a column named rent that contains rental prices, we can calculate the central tendency statistics listed above as follows:

```python
# Mean
rentals.rent.mean()
```

```python
# Median
rentals.rent.median()
```

```python
# Mode
rentals.rent.mode()
```

```python
# Trimmed mean
from scipy.stats import trim_mean
trim_mean(rentals.rent, proportiontocut=0.1) #trim extreme 10%
```

### Spread for numerical variable

The spread of a quantitative variable describes the amount of variability. This is important because it provides context for measures of central tendency.

There are several common measures of spread:

* Range
    * The difference between the maximum and minimum values of a variable.
* Interquartile range (IQR)
    * The difference between the 75th and 25th percentile values.
* Variance
    * The average of the squared distance from each data point to the mean.
* Standard deviation (SD)
    * The square root of the variance.
* Mean absolute deviation (MAD)
    * The mean absolute value of the distance between each data point and the mean.

To find range, IQR, variance, standard deviation, and MAD in python:

syntax:

```python
# Range
df_name.column_name.max() - df_name.column_name.min()

# IQR
df_name.column_name.quantile(0.75) - df_name.column_name.quantile(0.25)

#IQR (alternative way): using iqr() from scipy.stats
from scipy.stats from iqr
iqr(df_name.column_name)


# Variance
df_name.column_name.var()

# Standard deviation
df_name.column_name.std()
```

For our rentals DataFrame, we can calculate the spread for the rent column as follows:

```python
# Range
rentals.rent.max() - rentals.rent.min()
```

```python
# IQR 
rentals.rent.quantile(0.75) - rentals.rent.quantile(0.25)
```

```python
# IQR (alternative way)
from scipy.stats import iqr
iqr(rentals.rent)
```

```python
# Variance
rentals.rent.var()
```

```python
# Standard deviation
rentals.rent.std()
```

### Visualization of Numerical Data

For quantitative variables, boxplots and histograms are two common visualizations. 

These plots are useful because they simultaneously communicate information about:
* minimum and maximum values
* central location
* spread

Histograms can additionally illuminate patterns that can impact an analysis (e.g., skew or multimodality).

Python’s seaborn library, built on top of matplotlib, offers the boxplot() and histplot() functions to easily plot data from a pandas DataFrame:

To plot boxplot and histogram in python:

syntax:

```python
import matplotlib.pyplot as plt
import seaborn as sns

#boxplot
sns.boxplot(x="column_name", data=df_name)
plt.show()
plt.close()

#histogram
sns.histplot(x="column_name", data=df_name)
plt.show()
plt.close()
```

We need to use the plt function library from the matplotlib.pyplot to display the graph.

plt.show()- display the graph
plt.close()- close the window

Let's create a boxplot and histogram for the data column 'rent' in rentals dataset:

```python
import matplotlib.pyplot as plt
import seaborn as sns

#boxplot
sns.boxplot(x="rent", data=rentals)
plt.show()
plt.close()
```

```python
#histogram
sns.histplot(x="rent", data=rentals)
plt.show()
plt.close()
```

Both plots show that the distribution of rent is skewed to the right, with some outlier movies having extremely high budgets. 

This is consistent with the high mean rent value we saw earlier, since the mean is affected by skewness and outliers.

## Categorical Data Analysis

When it comes to categorical variables, the measures of central tendency and spread that worked for describing numeric variables, like mean and standard deviation, generally becomes unsuitable when we’re dealing with discrete values. 

Unlike numbers, categorical values are not continuous and oftentimes do not have an intrinsic ordering.

A good way to summarize categorical variables is to generate a table containing:

* frequency
* proportion

### Frequency for Cateogrical variable

The pandas library offers the .value_counts() method for generating the counts of all values in a DataFrame column:

syntax:

```python
df_name.column_name.value_counts()
```

By default, it returns the results sorted in descending order by count, where the top element is the mode, or the most frequently appearing value.

Example:

We are interested in the frequency(counts) of the rental listings in each borough in the rentals dataset:

```python
rentals.borough.value_counts()
```

### Proportion for Cateogrical variable

A counts table is one approach for exploring categorical variables, but sometimes it is useful to also look at the proportion of values in each category.

We can calculate the proportion for each category by dividing its count by the total number of values for that variable:

```python
df_name.column_name.value_counts() / len(df_name.column_name)
```

Or alternatively we could also obtain the proportions by specifying "normalize=True" to the .value_counts() method:

syntax:

```python
df.column_name.value_counts(normalize=True)
```

Example:

We are interested in the proportion of the rental listings in each borough in the rentals dataset:

```python
rentals.borough.value_counts(normalize=True)
```

### Visualization of Categorical Data

For categorical variables, 
* bar charts 
* pie charts 

are common options for visualizing the count (or proportion) of values in each category. 
They can also convey the relative frequencies of each category.

Python’s seaborn library offers several functions that can create bar charts. 

The simplest for plotting the counts is countplot():

syntax:

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Bar chart
sns.countplot(x="column_name", data=df_name)
plt.show()
plt.close()
```

To create pie chart, the pandas library provides a function around matplotlib‘s pie() function that can generate a pie chart from any column in a DataFrame:

syntax:

```python
import matplotlib.pyplot as plt

# pie chart
df_name.column_name.value_counts().plot.pie()
plt.show()
plt.close()
```

In general, many data analysts avoid pie charts because people are better at visually comparing areas of rectangles than wedges of a pie.

For a variable with a small number of categories (i.e., fewer than three), a pie chart is a reasonable choice; however, for more complex data, a bar chart is usually preferable.

Example:

We want to create a bar chart and pie chart for the relative frequency in each borough:

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Bar chart
sns.countplot(x="borough", data=rentals)
plt.show()
plt.close()
```

```python
import matplotlib.pyplot as plt

# pie chart
rentals.borough.value_counts().plot.pie()
plt.show()
plt.close()
```

From the plots, we can see that Manhattan borough appear most frequently and is the mode for rental listing. 

Queens borough appear least frequently in the dataset.
