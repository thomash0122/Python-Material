# Introduction of Data Cleaning in Exploratory Data Analysis

## Initial Data Inspection

Before analysis or cleaning, it is useful to print a few rows of data. This helps ensure that the data is properly loaded. 

It also allows us to compare the observed data to the data dictionary and determine whether the coding appears to match our expectations. 

We can do so with .head() method

## Data Information

Once we’ve taken a first look at some data, a common next step is to address questions such as:

* How many (non-null) observations do we have?
* How many unique columns/features do we have?
* Which columns (if any) contain missing data?
* What is the data type of each column?

we can easily address these questions using the .info() method

## Inspecting Missing Data

After identifying that there is some missing data and converting it to a format that Python can recognize, it’s often a good idea to take a closer look at those rows. 

Sometimes, we can find clues as to WHY the data is missing, which can help us make decisions about whether to get rid of the rows altogether or impute the missing values somehow.

To perform such finding missing data:

syntax:

```python
df_name[df_name.isnull().any(axis=1)]
```

This will return every corresponding row data that contain missing value in any columns

### Inspecting Missing Data

Since there are some 0 in some values in Glucose, BloodPressure, SkinThickness, Insulin, BMI columns where it shouldn't, we decided to treat them as missing data.
Now we should replace 0s in those column with NaN to represent missing value

```python
df[["Glucose", "BloodPressure", "SkinThickness","Insulin", "BMI"]] = df[["Glucose", "BloodPressure", "SkinThickness","Insulin", "BMI"]].replace(0, np.nan)
```

Now we should examine how many missing value are there in each column:

```python
df.isna().sum()
```

Now, we dive deep into those data rows with mising values. Let's print out the row data with missing values:

```python
df[df.isna().any(axis=1)]
```

From all the data missing rows:

* Majority of the data rows has more than 1 missing values
* The values in "Insulin" column are all NaN
    * This directly tells us those patient didnt have their Insulin recorded
    * further update is required for this dataset

### Variable Type Analysis:

Let's take a look at the current datatype of each columns:

We see that the "Outcome" column is represented as integer instead of categorical. WE should convert it from integer type to catgeorical type.

```python
df["Outcome"] = pd.Categorical(
    df["Outcome"],
    [0,1],
    ordered=True
)
```
