This table tells us the number of people who gave each possible combination of responses to these two questions.

To assess whether there is an association between these two variables, we need to ask whether information about one variable gives us information about the other. 

In this example, we see that among people who don’t see themselves as a leader (the first column), a larger number (3015) don’t think they have a talent for influencing people. So, if we know how someone responded to the leadership question, we have some information about how they are likely to respond to the influence question. This suggests that the variables are associated.

## Marginal Proportions

## Chi-Sqaure Statisitc

To examine if 2 categorical variable are assoicate to each other or not (Test for independence of 2 categorical variable): we must perform hypothesis testing.

First steps- Null and Alternate Hypothesis:

* Null Hypothesis: 2 categorical variables are independent (associated)
* Alternate Hypothesis: 2 categorical variables are not independent (not associated)

Second steps- Calculate chi-sq statistic information:

The chi-sq test statistics can be obtain from chi2_contingency functon in scipy.stats:

```python
import pandas as pd
from scipy.stats import chi2_contingency

#Create a frequency table
var_name_freq = pd.crosstab(df_name.column_name1, df_name.column_name2)

#Find chi-sq test statistic
chi2, p, df, expect_table = chi2_contingency(var_name_freq)
```

The chi2_contingency() function has 4 output:
* chi-sq test statistic
* p-value
* degree of freedom
* expected contingency table

Third steps- Conclusions:

If the p-value is lower than the confidence level(alpha value)
* we reject the null hypothesis, and conclude that 2 categorical variable are not independent (not associated)

If the p-value is higher than the confidence level(alpha value)
* we reject the null hypothesis, and conclude taht 2 categorical variable are not independent (not associated)

Example: Suppose we want to test if leader and influence is assoicated or not?
