# Bar plot

The plt.bar function allows you to create simple bar charts to **compare multiple categories of data**.

Some possible data that would be displayed with a bar chart:

* x-axis: famous buildings, y-axis: heights
* x-axis: different planets, y-axis: number of days in the year
* x-axis: programming languages, y-axis: lines of code written by you

Syntax:

```python
plt.bar(x_values, y_values)
```

You call plt.bar with two arguments:

* the x-values — a list of x-positions for each bar
* the y-values — a list of heights for each bar

In most cases, we will want our x-values to be a list that looks like [0, 1, 2, 3 ...] and has the same number of elements as our y-values list.

To do so, we can use range() function to create the integer that represents each categories

Example:

```python
import matplotlib.pyplot as plt

drinks = ["cappuccino", "latte", "chai", "americano", "mocha", "espresso"]
sales =  [91, 76, 56, 66, 52, 27]

x_values = range(len(drinks))

plt.bar(x_values, sales)
plt.show()
```

## Modify ticks

Now that we construct the carplot, but we are representing the categories with integers. So we need to modify the ticks so that we can label each tick as categories.

Syntax:

```python
#Create ax object
ax = plt.subplot()

#Create plot
plt.bar(x_values,y_values)

#Modify x ticks
ax.set_xticks([...])

#Modify x tick labels
ax.set_xticklabels([...])
```

Example:

```python
import matplotlib.pyplot as plt

drinks = ["cappuccino", "latte", "chai", "americano", "mocha", "espresso"]
sales =  [91, 76, 56, 66, 52, 27]

x_values = range(len(drinks))

ax = plt.subplot()
plt.bar(x_values, sales)

ax.set_xticks(range(len(drinks)))
ax.set_xticklabels(drinks)
plt.show()
```

## Stacked Bars

If we want to compare two sets of data while preserving knowledge of the total between them, we can also stack the bars.

To do so, we need to first plot the bottom bar plot and then plot the top bar plot

Syntax

```python
#Top bar plot
plt.bar(x_values, y_values_bottom)

#Bottom bar plot
plt.bar(x_values, y_values_top, bottom=y_values_bottom)
```

Example:

```python
drinks = ["cappuccino", "latte", "chai", "americano", "mocha", "espresso"]
sales1 =  [91, 76, 56, 66, 52, 27]
sales2 = [65, 82, 36, 68, 38, 40]

ax = plt.subplot()
plt.bar(range(len(drinks)), sales1)
plt.bar(range(len(drinks)), sales2, bottom=sales1)
plt.legend(["Location 1", "Location 2"])
ax.set_xticks(range(len(drinks)))
ax.set_xticklabels(drinks)
plt.show()
```

## Error Bars

In the previous exercise, you learned to represent data as bars of different heights. Sometimes, we need to visually communicate some sort of uncertainty in the heights of those bars.

To display error bar,

Syntax:

```python
plt.bar(x_values, y_values, yerr=..., capsize=...)
```

Argument:
* yerr: represents the uncertainty
    * integer- if the uncertainty is the same across the bar
    * list- if the uncerstainty is different across the bar
* capsize: represents the cap wide

Example:

```python
drinks = ["cappuccino", "latte", "chai", "americano", "mocha", "espresso"]
ounces_of_milk = [6, 9, 4, 0, 9, 0]
error = [0.6, 0.9, 0.4, 0, 0.9, 0]

ax=plt.subplot()
plt.bar(range(len(drinks)), ounces_of_milk, yerr=error, capsize=10)
ax.set_xticks(range(len(drinks)))
ax.set_xticklabels(drinks)
plt.show()
```

There are 2 kind of Categorial data
* Ordinal

* Nominal
    * The kind of data where there is no specific order
