# Histogram

A histogram is a chart that plots the distribution of a numeric variable's values as a series of bars. 

Each bar typically covers a range of numeric values called a bin or class; a bar's height indicates the frequency of data points with a value within the corresponding bin.

Syntax:

```python
import matplotlib.pyplot as plt

plt.hist(dataset, range=(low,high), bins=num)
```

Arugment:
* **range**: The keyword range selects the minimum and maximum values to plot
* **bins**: The keyword bins to set how many bins we want to divide the data into

Example:

```python
import numpy as np

dataset = np.random.normal(0,20,1000)

plt.hist(dataset)
plt.show()
plt.close()
```

We want to take the subset of the sample N(0,20) distribution from (-60,0) and divide into 40 bins

```python
plt.hist(dataset, range=(-60,0), bins=40)
plt.show()
plt.close()
```

## Multiple Histogram

If we want to compare two different distributions, we can put multiple histograms on the same plot. 

But plotting 2 histogram on the same plot is hard to read on top of each other, so we can either:

* use alpha keyword, set the transparency
* use the histype with the argument "step to draw just the outline of the histogram

Example:

```python
a = np.random.normal(loc=64, scale=2, size=10000)
b = np.random.normal(loc=70, scale=2, size=100000)

plt.hist(a)
plt.hist(b)
plt.show()
plt.close()
```

## Standardized Distribution

The result above is two histograms that are very difficult to compare. 

To solve this, we can normalize our histograms using density=True. This command divides the height of each column by a constant such that the total shaded area of the histogram sums to 1.

Syntax:

```python
plt.hist(dataset, density=True)
```

Example:

```python
a = np.random.normal(loc=64, scale=2, size=10000)
b = np.random.normal(loc=70, scale=2, size=100000)

plt.hist(a, density=True, alpha=0.2)
plt.hist(b, density=True, alpha=0.2)
plt.show()
plt.close()
```
