# Aggregate Statistics

In this lesson, you will learn about aggregates in Pandas. 

An aggregate statistic is a way of creating a single number that describes a group of numbers. Common aggregate statistics include mean, median, and standard deviation.

You will also learn how to rearrange a DataFrame into a pivot table, which is a great way to compare data across two dimensions.

## Calculating Column Statistics

Aggregate functions summarize many data points (i.e., a column of a dataframe) into a smaller set of values.

The following table summarizes some common commands:

* mean()- Average of all values in column

* std()- Standard deviation

* median()-	Median

* max()- Maximum value in column

* min()- Minimum value in column

* count()- Number of values in column

* nunique()- Number of unique values in column

* unique()-	List of unique values in column

* sum() - Sum of all the values in column

To use those commands:

syntax:

```python
var_name = df_name.column_name.aggregate_function()
```

Example:

Given a DataFrame

```python
import pandas as pd

grades = pd.DataFrame({
    "student": ["Amy", "Amy", "Bob", "Bob", "Chris", "Chris"],
    "assignment_name": ["Assignment1", "Assignment2", "Assignment1", "Assignment2", "Assignment1", "Assignment2"],
    "grade":[75, 35, 99, 35, 85, 90]
})
```

```python
print(grades)
```

If we want to know the average of the overall student's grade in all assignment:

```python
average = grades.grade.mean()
print(average)
```

### Calculating Aggregate Functions 

When we have a bunch of data, we often want to calculate aggregate statistics (mean, standard deviation, median, percentiles, etc.) over certain subsets of the data.

To apply aggregate function of each group of data, we can:

syntax:

```python
df_name.groupby("column_name1").column_name2.aggregate_function()
```

where:

* column_name1 is the column that we want to group by

* column_name2 is the column that we want to aggregate on 

* aggregate_function is the aggregate function we want to apply 

Example:

Given a DataFrame:

```python
import pandas as pd

grades = pd.DataFrame({
    "student": ["Amy", "Amy", "Bob", "Bob", "Chris", "Chris"],
    "assignment_name": ["Assignment1", "Assignment2", "Assignment1", "Assignment2", "Assignment1", "Assignment2"],
    "grade":[75, 35, 99, 35, 85, 90]
})
```

```python
print(grades)
```

Suppose we want to know the all student's average grade of each assignment:

```python
average_each_assignment = grades.groupby("assignment_name").grade.mean()
```

```python
print(average_each_assignment)
```

Notice what datatype is the result:

```python
print(type(average_each_assignment))
```

The groupby function creates a new Series, not a DataFrame. 

A Series is a one-dimensional array of data. 

It can hold data of any type: string, integer, float, dictionaries, lists, booleans, and more.

The easiest way to conceptualize a Series is a single column in a table, which is why it's considered one-dimensional.

Usually, we’d prefer that those indices were actually a column. In order to get that, we can use reset_index()at the end of the aggregate function. 

This will transform our Series into a DataFrame and move the indices into their own column.

syntax:

```python
var_name = df_name.groupby("column_name1").column_name2.aggregate_function().reset_index()
```

Example:

Given a DataFrame:

```python
import pandas as pd

grades = pd.DataFrame({
    "student": ["Amy", "Amy", "Bob", "Bob", "Chris", "Chris"],
    "assignment_name": ["Assignment1", "Assignment2", "Assignment1", "Assignment2", "Assignment1", "Assignment2"],
    "grade":[75, 35, 99, 35, 85, 90]
})
```

```python
print(grades)
```

Suppose we want to know the average of all assignment of each student and store it in a dataframe:

```python
average_each_student = grades.groupby("student").grade.mean().reset_index()
print(average_each_student)
```

```python
print(type(average_each_student))
```

Now, the aggregated result is save as a DataFrame instead of a Series

### Aggregate function with lambda function

Instead of applying the common aggregate function, we can apply lambda function as aggregate function as well

Note that the input to our lambda function will always be a list of values.

Example:

Give a DataFrame:

```python
import pandas as pd

grades = pd.DataFrame({
    "student": ["Amy", "Amy", "Bob", "Bob", "Chris", "Chris"],
    "assignment_name": ["Assignment1", "Assignment2", "Assignment1", "Assignment2", "Assignment1", "Assignment2"],
    "grade":[75, 35, 99, 35, 85, 90]
})
```

```python
print(grades)
```

If we want to find the 75 percentile of the grade of each Assignment from all the student:

```python
import numpy as np

third_quartile = grades.groupby("assignment_name").grade.apply(lambda list: np.percentile(list, 75)).reset_index()
print(third_quartile)
```

np.percentile( array_like_object, number_of_percentile )

* 1st argument: array_like_object- can be a list

* 2nd argument: number_of_percentile- indicate the percentile you want to get

Remember, since percentile() is a function form the module "numpy" so we have to import the "numpy" module first

## Pivot Tables

The pivot() function is an incredibly useful tool for transforming and summarizing data. 

It allows you to restructure a DataFrame by turning rows into columns and columns into rows based on a specified index column, a specified columns column, and a specified values column.

For instance:

We suspected that there might be different sales on different days of the week at different stores, so we performed a `groupby` across two different columns (`Location` and `Day of Week`). This gave us results that looked like this: 

| Location | Day of Week | Total Sales |
| --- | --- | --- |
| Chelsea | M | 300 |
| Chelsea | Tu | 310 |
| Chelsea | W | 375 |
| Chelsea | Th | 390 |
|...|
| West Village | Th | 450 |
| West Village | F | 390 |
| West Village | Sa | 250 |
|...|		

In order to test our hypothesis, it would be more useful if the table was formatted like this:

Location	    M	Tu	W	Th	F	Sa	Su
Chelsea	        300	310	375	390	300	150	175
West Village	300	310	400	450	390	250	200
…

Reorganizing a table in this way is called pivoting. The new table is called a pivot table.

To create a pivot table:

syntax:

```python
var_name = df_name.pivot(
    columns = "ColumnToPivot",
    index = "ColumnToBeRows",
    values = "ColumnToBeValues"
).reset_index()
```

Just like with groupby, the output of a pivot command is a new DataFrame, but the indexing tends to be “weird”, so we usually follow up with .reset_index().
