# Python Pandas- Modifying DataFrames

## Adding a Column 

To add a new column to an existing DataFrame, there are 3 different way to do it:

* specify the row data in a list

* add a column that is the same for all rows

* add a column by perform a function on the existing columns

### Specify the row data in a list

One way that we can add a new column is by giving a list of the same length as the existing DataFrame.

syntax:

```python
df["new_column_name"] = ["value1", "value2", "value3", ...]
```

Example:

Given a DataFrame:

```python
import pandas as pd

df = pd.DataFrame([
  [1, '3 inch screw', 0.5, 0.75],
  [2, '2 inch nail', 0.10, 0.25],
  [3, 'hammer', 3.00, 5.50],
  [4, 'screwdriver', 2.50, 3.00]
],
  columns=['Product ID', 'Description', 'Cost to Manufacture', 'Price']
)
```

```python
print(df)
```

To add a column to df called 'Sold in Bulk?', which indicates if the product is sold in bulk or individually

```python
df["Sold in Bulk"] = ["Yes", "Yes", "No", "No"]
```

The DataDrame after adding new column:

```python
print(df)
```

### Add a column that is the same for all rows

We can also add a new column that is the same for all rows in the DataFrame.

syntax:

```python
df["new_column_name"] = value
```

Example:

Given a DataFrame:

```python
import pandas as pd

df = pd.DataFrame([
  [1, '3 inch screw', 0.5, 0.75],
  [2, '2 inch nail', 0.10, 0.25],
  [3, 'hammer', 3.00, 5.50],
  [4, 'screwdriver', 2.50, 3.00]
],
  columns=['Product ID', 'Description', 'Cost to Manufacture', 'Price']
)
```

```python
print(df)
```

Add a column to df called Is taxed?, which indicates whether or not to collect sales tax on the product. It should be 'Yes' for all rows.

```python
df["Is taxed?"] = "Yes"
```

The DataFrame after adding new column:

```python
print(df)
```

### Perform a function on the existing columns

you can add a new column by performing a function on the existing columns.

syntax:

```python
df["new_column_name"] = function(df.existing_column_name)
```

Example:

Given a DataFrame:

```python
import pandas as pd

df = pd.DataFrame([
  [1, '3 inch screw', 0.5, 0.75],
  [2, '2 inch nail', 0.10, 0.25],
  [3, 'hammer', 3.00, 5.50],
  [4, 'screwdriver', 2.50, 3.00]
],
  columns=['Product ID', 'Description', 'Cost to Manufacture', 'Price']
)
```

```python
print(df)
```

Add a column to df called 'Margin', which is equal to the difference between the Price and the Cost to Manufacture.

```python
df["Margin"] = df.Price - df["Cost to Manufacture"]
```

Be Careful!!! 

You’ll need to access the values of Cost to Manufacture using df['Cost to Manufacture'] not df.Cost of Manufacture because the column name doesn’t follow our rules for variable names.

The DataFrame after adding new column:

```python
print(df)
```

### Applying Lambda Function to a Column

Let's recall what lambda function is:

In Python, a lambda function is a one-line shorthand for function. A simple lambda function might look like this:

syntax:

```python
function_var = lambda input: <WHAT_TO_RETURN_IF_STATEMENT_IS_TRUE> if <IF_STATEMENT> else <WHAT_TO_RETURN_IF_STATEMENT_IS_FALSE>
```

In Pandas, we often use lambda functions to perform complex operations on columns.

syntax:

```python
df["new_column_name"] = df.existing_column.apply(lambda_function)
```

Example:

Given a DataFrame:

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["JOHN SMITH", "Jane Doe", "joe schmo"],
    "Email": ["john.smith@gmail.com", "jdoe@yahoo.com", "joeschmo@hotmail.com"]
})
```

```python
print(df)
```

Suppose that we want to create a column containing the email provider for each email address:

First, we must define a lambda function to find the email provider.

```python
find_email_provider = lambda str: str.split("@")[-1]
```

```python
df["Email Provider"] = df.Email.apply(find_email_provider)
```

DataFrame after adding a column:

```python
print(df)
```

Note: If the lambda function code got too long

For example:

```python
total_earned = lambda row: \
  row.hourly_wage * 40 + (row.hours_worked-40)*row.hourly_wage*1.5 \
  if row.hours_worked > 40 \
  else row.hours_worked * row.hourly_wage
```

We need use "\"- line continuation and the string will continue on the next line as if it were a single line. because lambda function is a SINGLE LINE command.

### Applying Lambda Function to a row

We can also operate on multiple column to get our desired row data.

If we use apply without specifying a single column and add the argument axis=1, the input to our lambda function will be an entire row, not a column.

Example:

Given a DataFrame:

```python
import pandas as pd

df = pd.DataFrame({
    "Item": ["Apple", "Milk", "Paper Towels", "Light Bulbs"],
    "Price": [1.00, 4.20, 5.00, 3.75],
    "Is taxed?": ["No", "No", "Yes", "Yes"]
})
```

```python
print(df)
```

If we want to add in the price with tax (7.5% sales tax) for each line, we’ll need to look at two columns: Price and Is taxed?.

First, we need to create a lambda function to find the price with tax usng 2 columns: Price and Is taxed?

```python
taxed = lambda row: row.Price * 1.075 if row["Is taxed?"] == "Yes" else row.Price
```

```python
df["Price with Tax"] = df.apply(taxed, axis=1)
```

DataFrame after adding a column:

```python
print(df)
```

## Rename Column

To change the name of the column, there are 2 ways of doing this:

* using .columns to respecify the column name using list

* using .rename method

### use .columns to respecify

syntax:

```python
df_name.columns = ["new_column1_name", "new_column2_name", ...]
```

Example:

Give a DataFrame:

```python
import pandas as pd

df = pd.DataFrame({
    'name': ['John', 'Jane', 'Sue', 'Fred'],
    'age': [23, 29, 21, 18]
})
```

```python
print(df)
```

To change from "name" to "First Name" and "age" to "Age"

```python
df.columns = ["First Name", "Age"]
```

DataFrame after renaming the columns:

```python
print(df)
```

This is great when you need to change all of the column names at once. 

### .rename method

syntax:

```python
df.rename(columns = {
    current_column1_name : new_column1_name,
    current_column2_name : new_column2_name,
    ...
    },
    inplace = True        
)
```

The inplace = True is used to indicate that we are modify on the existng DataFrame

Example:

Given a DataFrame:

```python
df = pd.DataFrame({
    'name': ['John', 'Jane', 'Sue', 'Fred'],
    'Age': [23, 29, 21, 18],
    'Sex': ["Male", "Female", "Female", "Male"]
})
```

```python
print(df)
```

To change only from "name" to "First Name"

```python
df.rename(columns={
    "name": "First Name"
    },
    inplace=True
)
```

DataFrame after renaming the columns:

```python
print(df)
```

There are several reasons why .rename is preferable to .columns:

* You can rename just one column

* You can be specific about which column names are getting changed (with .column you can accidentally switch column names if you’re not careful)

Note: If you misspell one of the original column names, this command won’t fail. It just won’t change anything.
