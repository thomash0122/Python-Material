## Diagnose the Data

We often describe data that is easy to analyze and visualize as **tidy data**. 

What does it mean to have tidy data?

For data to be tidy, it must have:

* **Each variable as a separate column**
* **Each row as a separate observation**

The first step of diagnosing whether or not a dataset is tidy is **using pandas functions to explore and probe the dataset**.

Some useful dataset diagnostic:

|Code|Descriptions|
|---|---|
|.head()| display the first 5 rows of the table|
|.info() |display a summary of the table|
|.describe()| display the summary statistics of the table|
|.columns| display the column names of the table|
|.value_counts() |display the distinct values for a column|

Example:

Given a data:

```python
students = pd.DataFrame({
    "full_name":["Moses Kirckman","Timofei Strowan","Silvain Poll","Lezley Pinxton","Bernadene Saunper"],
    "gender_age":["M14","M14","M14","M15","F17"],
    "fractions":["76%","83%","83%","80%","67%"],
    "probability":["72%",np.nan,np.nan,"86%","78%"],
    "grade":["9th grade","12th grade","12th grade","11th grade","10th grade"]
})
```

```python
#first 5 rows of the table
students.head()
```

```python
#summary of the table
students.info()
```

```python
#summary statistics of the table
students.describe()
```

```python
#column names of the table
students.columns
```

```python
#distinct values for each column
for column in students.columns:
    print(students[column].value_counts())
    print("\n")
```

To reshape the data using pandas:

**Syntax:**

```python
df_name = pd.melt(
    frame = "df_name",
    id_vars = ["data_columns"...],
    value_vars = ["data_columns"...],
    value_name = "df_column",
    var_name = "df_column"
)
```

The parameters you provide are:

|arugment|description|
|---|---|
|frame| the DataFrame you want to melt|
|id_vars| the column(s) of the old DataFrame to preserve|
|value_vars| the column(s) of the old DataFrame that you want to turn into variables|
|value_name| what to call the column of the new DataFrame that stores the values|
|var_name| what to call the column of the new DataFrame that stores the variables|

Example:

To reshape the students dataset

```python
students = pd.melt(
    frame = students,
    id_vars = ["full_name","gender_age","grade"],
    value_vars = ["fractions","probability"],
    value_name = "score",
    var_name = "exam"
)
```

## Splitting by Index

In the students dataset, we see the "gender_age" data column contains 2 kind of information:

   * gender
   * age
   
We want to split this data into gender and age so that we can use these columns as separate features.

We can use the **pd.Series.str** to access the values of the series as strings and apply several methods to it.

To split by index

**Sytnax:**

```python
df_name["df_new_column_name"] = df_name["df_old_column_name"].str[index]
```

Example:

We want to split the "gender_age" data column in students dataset into 2 separate columns:
* gender
* age

```python
students.gender_age
```

```python
#gender is indicated by M/F
students["gender"] = students.gender_age.str[0]
```

```python
#age is indicated by the number after gender
students["age"] = students.gender_age.str[1:]
```

```python
#drop the gender_age data column
students.drop(
    columns=["gender_age"],
    inplace=True
)
```

```python
students
```

## String Parsing using Regex

Sometimes we need to modify strings in our DataFrames to help us transform them into more meaningful metrics.

In the students example, score is an object datatype, but we can't directly convert score from object type to numeric type **since score has a "$" character that prevent from converting**

we need to remove the "%" character from score

To remove character from each entry, we can use the .replace() method and use regex to represent the string character

**Syntax:**

```python
df_name["column_name"] = df_name.column_name.replace("[charactor_code]", "replace_by", regex=True)
```

replace() method:
* first agrument: represent the string you trying to replace in charactor code
* second agrument: represent the string that you trying to replace by
* third agrument: enable us to use regex

Example:

we need to remove the "%" character from score

```python
students["score"] = students.score.replace("\%"," ", regex=True)
```

```python
students
```

Now we can convert "score column" into numeric datatype

```python
students["score"] = pd.to_numeric(students.score)
```

```python
students.dtypes
```

Example:

We want to modify "grade" column so that there is only numeric component 

```python
#define a series with each entry splited by white space
grade_split = students.grade.str.split(" ")
```

```python
grade_split
```

```python
#set the "grade" column to the Oth index of grade_split
students["grade"] = grade_split.str.get(0)
```

```python
students
```

Now we need to remove the **"th"** in each entry of the "grade". We can do so by **replacing the non-numeric component with whitespace**

```python
students["grade"] = students.grade.replace("[^\d]"," ",regex=True)
```

```python
students
```

Now we can convert "grade" column from object type to numeric type

```python
students["grade"] = pd.to_numeric(students.grade)
```

```python
students
```

```python
students.dtypes
```

## Missing Value

We often have data with missing elements, as a result of a problem with the data collection process or errors in the way the data was stored. 

The missing elements normally show up as NaN (or Not a Number) values

To deal with missing value:
* drop all of the rows with a missing value
* fill the missing values with mean or some aggregated value

#### To drop all of the rows with missing value, we can use the **.dropna()** method.

**Syntax:**

```python
df_name = df_name.dropna()
```

If we want to remove every row with Nan value in the specific column, we can specify the "subset" argument

**Syntax:**

```python
df_name = df_name.dropna(subset=["column_name"...])
```

To check if there is any NaN value in the dataframe:

**Syntax:**

```python
#return a boolean table
df_name.isna()

#return number of NaN value 
df_name.isna().sum()
```

Example:

We want to drop the NaN value in all column in students dataset

```python
students_no_nan = students.dropna()
```

```python
#return a boolean table
students_no_nan.isna()
```

```python
#return number of NaN value 
students_no_nan.isna().sum()
```

#### To fill all of the missing value with aggregated data, we can use the .fillna() method

**Syntax:**

```python
df_name = df_name.fillna(value={
    "column_name1":aggregated_data1,
    "column_name2":aggregated_data2,
    ...
})
```

Example:
