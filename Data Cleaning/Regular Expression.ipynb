# Regular Expression

Regular expressions, often shortened to regex or regexp, is a language used for pattern-matching text content.

A regular expression pattern is a string made up of normal characters mixed with metacharacters that have special meanings defining how characters should be matched.

For instance, 

The pattern “h” by itself would match the single letter “h

Whereas if we add the metacharacter “.”, which matches any one character, then “h.” would match sequences such as “ha”, “ho”, “h5” and “h!”.

## Literals

The simplest text we can match with regular expressions are literals. 

This is where our regular expression contains the exact text that we want to match.

Example:

"**bark**" match with **bark**


## Alternation

Alternation, performed in regular expressions with the pipe symbol, **|**, allows us to match either the characters preceding the **|** OR the characters after the **|**. 

**Matches the pattern on the right of the | or the pattern on the left.**

Example:

"I love baboons **|** I love gorillas"

match with 

I love **baboons**

or 

I love **gorillas**

## Character Sets

Character sets, denoted by a pair of brackets **[ ]**, let us match one character from a series of characters, allowing for matches with incorrect or different spellings.

**Matches the characters inside the brackets.**

Example:

"**[bcm]** at"

matches with
* **c**at

and

* **b**at

and

* **m**at

## Wildcards

Wildcards will match any single character (letter, number, symbol or whitespace) in a piece of text. 

**Matches any one character**

Example:

".at"

matches with any one character and "at" after it

## Ranges

Ranges allow us to specify a range of characters in which we can make a match without having to type out each individual character. 

**Matches the characters inside the brackets. Can also specify ranges of characters such as [a-z].**

* To specify 0 to 9: **[0-9]**
* To specify a to z (lower case): **[a-z]**
* To specify A to Z (Upper case): **[A-Z]**
* To specify all alphabet (including lower and upper): **[A-Za-z]**

Example:

"I adopted **[2-9]** **[b-h]** ats"

will match the text 
* I adopted 4 bats
* I adopted 8 cats
* I adopted 5 hats
etc...

## Shorthand Character Classes

Shorthand Character Classes used represent common ranges.

|Classes|Description|
|---|---|
|\w|The “word character” class represents the regex range **[A-Za-z0-9_]**, and it matches a single uppercase character, lowercase character, digit or underscore|
|\d| The “digit character” class represents the regex range **[0-9]**, and it matches a single digit character|
|\s|the “whitespace character” class represents the regex range **[ \t\r\n\f\v]**, matching a single space, tab, carriage return, line break, form feed, or vertical tab|


Example:

"**\d\s\w\w\w\w\w\w\w**"

Matches a digit character, followed by a whitespace character, followed by 7 word characters. Thus the regex completely matches the text 3 monkeys.

## Negated shorthand character classes

These shorthands will match any character that is NOT in the regular shorthand classes.

|Classes|Description|
|---|---|
|\W| the “non-word character” class represents the regex range **[^A-Za-z0-9_]**, matching any character that is not included in the range represented by \w|
|\D| the “non-digit character” class represents the regex range **[^0-9]**, matching any character that is not included in the range represented by \d|
|\S| the “non-whitespace character” class represents the regex range **[^ \t\r\n\f\v]**, matching any character that is not included in the range represented by \s|

## Grouping

Similar to Alteration, grouping is very useful in limiting alternation to part of the regex.

Example:

In Alternation, "I love baboons **|** I love gorillas" matches with
* I love baboons

or

* I love gorillas

To represent the same regex as the example above using Grouping:

"I love **(baboons|gorillas)**"

## Quantifiers-Fixed 

Fixed quantifiers, denoted with curly braces {}, let us indicate the exact quantity of a character we wish to match, or allow us to provide a quantity range to match on.

**Matches the preceding element exactly n times.**

Example:

* **\w{3}** match exactly 3 word characters
* **\w{4,7}** match at minimum 4 and maximum 7 word characters
* "roa **{3}** r" match the characters "ro" followed by 3 "a"s and the character r

## Quantifiers-Optional

Optional quantifiers, indicated by the question mark ?, allow us to indicate a character in a regex is optional, or can appear either 0 times or 1 time. 

Example:

"humou **?** r" match with 
* humor

or

* humour

"The monkey ate a **(rotten )?** banana" match with
* The monkey ate a rotten banana
* The monkey ate a banana

## Quantifier - 0 or More

The Kleene star, denoted with the asterisk *, is also a quantifier, and matches the preceding character 0 or more times. 

This means that the character **doesn’t need to appear, can appear once, or can appear many many times.**

Example:

**"meo*w"**

Matches 
* **the characters "me"**
* **followed 0 or more "o"s**
* **followed by a w**

ex) mew, meow, mooow, meoooooooooooooooooooow

## Quantifier - 1 or More

Kleene plus, denoted by the plus **+**, which matches the preceding character 1 or more times.

Similar to Kleene star, but the **character appears at least once**

Example

"meo+w"

Matches 
* **the characters "me"**
* **followed 1 or more "o"s**
* **followed by a w**

ex) meow, mooow, meoooooooooooooooooooow, but not match mew

## Anchors

The anchors hat ^ and dollar sign $ are used to match text at the start and the end of a string, respectively.

Example:

"**^**Monkeys: my mortal enemy **$**"

Matches completely with the text:
* "**Monkeys: my mortal enemy**"

But not match: 
* Spider **Monkeys: my mortal enemy** in the wild
* Squirrel **Monkeys: my mortal enemy** in the wild

The ^ ensures that the matched text begins with Monkeys, and the $ ensures the matched text ends with enemy.
